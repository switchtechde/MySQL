```python
import json 
file_path = "C:\\Users\\joshi\\OneDrive\\Desktop\\MySQL\\PySpark\\Pyspark details\\Data\\Financial Transactions\\mcc_codes.json"

with open(file_path, "r") as f:
    mcc_codes = json.load(f)

data = [
    (int(mcc_code),description) 
    for mcc_code, description in mcc_codes.items()
]
```

```python
df = spark.createDataFrame(data, ["mcc_code", "description"])
df.show(truncate=False)
```
